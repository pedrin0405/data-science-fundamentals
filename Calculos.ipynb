```python
def somantorio_ao_quadrado(n):
    result = 0
    for i in range(n + 1):
        result = result + i**2
    return result

n = 5
print(somantorio_ao_quadrado(n))
```

```python
def soma_dados(conjunto):
  soma = 0
  for dado in conjunto:
    soma += dado
  return soma

conjunto = [12, 15, 9, 14, 10]
print(soma_dados(conjunto))
```

```python
def media_aritmetica(conjunto):
  soma = 0
  for dado in conjunto:
    soma += dado
  return soma / len(conjunto)

conjunto1 = [5,7,8,10]
conjunto2 = [2, 4, 6, 8, 10]
conjunto3 = [12, 15, 9, 14, 10]
print(f'A) {conjunto1} = {media_aritmetica(conjunto1)}')
print(f'B) {conjunto2} = {media_aritmetica(conjunto2)}')
print(f'C) {conjunto3} = {media_aritmetica(conjunto3)}')
```

```python
def soma_dados_ao_quadradro(conjunto):
  soma = 0
  for dado in conjunto:
    soma += dado ** 2
  return soma

conjunto = [1, 2, 3, 4, 5, 6]
print(soma_dados_ao_quadradro(conjunto))
```

```python
def media_harmonica(conjunto):
    soma_reciprocos = 0
    for dado in conjunto:
        soma_reciprocos += 1 / dado
    return len(conjunto) / soma_reciprocos

conjunto1 = [4, 6]
conjunto2 = [3, 5, 15]
print(f'A) {conjunto1} = {media_harmonica(conjunto1)}')
print(f'B) {conjunto2} = {media_harmonica(conjunto2)}')
```

```python
def media_geometrica(conjunto):
    produto = 1
    for dado in conjunto:
        produto *= dado
    return produto ** (1 / len(conjunto))

conjunto1 = [2, 8]
conjunto2 = [1.2, 1.10, 0.95]
print(f"{media_geometrica(conjunto1):.2f}")
print(f"{media_geometrica(conjunto2):.2f}")
```

```python
def media_quadratica(conjunto):
    soma_quadrados = 0
    for dado in conjunto:
        soma_quadrados += dado ** 2
    return (soma_quadrados / len(conjunto)) ** 0.5

conjunto1 = [3, 4]
conjunto2 = [2, 5, 6]
print(f"{media_quadratica(conjunto1):.2f}")
print(f"{media_quadratica(conjunto2):.2f}")
```

```python
def media_ponderada(valores, pesos):
    soma_produtos = 0
    soma_pesos = 0
    for i in range(len(valores)):
        soma_produtos += valores[i] * pesos[i]
        soma_pesos += pesos[i]
    return soma_produtos / soma_pesos

valores = [6, 8, 9]
pesos = [2, 3, 1]

print(media_ponderada(valores, pesos))
```

```python
def mediana(conjunto):
    dados = sorted(conjunto)
    n = len(dados)
    if n == 0:
        return None
    meio = n // 2
    if n % 2 == 1:
        return dados[meio]
    else:
        return (dados[meio - 1] + dados[meio]) / 2

conjunto_impar = [15, 12, 9, 14, 10]
conjunto_par = [7, 1, 3, 9, 2, 5]

print(f"Mediana (ímpar): {mediana(conjunto_impar)}")
print(f"Mediana (par): {mediana(conjunto_par)}")
```

```python
def moda(conjunto):
    if not conjunto:
        return None
    
    frequencias = {}
    for dado in conjunto:
        frequencias[dado] = frequencias.get(dado, 0) + 1
    
    max_freq = max(frequencias.values())
    
    # Se nenhum elemento se repete (todos freq == 1), o conjunto é amodal
    if max_freq == 1 and len(conjunto) > 1:
        return "Amodal (sem repetições)"
    
    modas = [dado for dado, freq in frequencias.items() if freq == max_freq]
    return modas if len(modas) > 1 else modas[0]

# Testes com conjuntos de dados
conjunto1 = [2, 3, 5, 3, 7, 3, 9]
conjunto2 = [1, 2, 2, 3, 4, 4, 5]
conjunto3 = [10, 20, 30, 40]

print(f"Unimodal {conjunto1}: {moda(conjunto1)}")
print(f"Bimodal {conjunto2}: {moda(conjunto2)}")
print(f"Amodal {conjunto3}: {moda(conjunto3)}")
```
